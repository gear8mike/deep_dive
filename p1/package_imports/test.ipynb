```python
from collections import namedtuple

def default_namedtuple(class_name,**fields):
    Struct = namedtuple('Struct', fields.keys())
    Struct.__new__.__defaults__ = tuple(fields.values())
    return Struct
```

```python
Vector2D = default_namedtuple(class_name='Vector2D',
                              x1=None, y1=None,
                              x2=None, y2=None,
                              x0=0, y0=0)
```
